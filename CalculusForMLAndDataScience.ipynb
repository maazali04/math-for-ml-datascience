## Limit
A limit is the value that a function or sequence approaches as the input (or index) gets closer and closer to a specific number.

Remeber: For limit to exists, right side limit must equal to left side limit.


e.g. Find the limit of $\frac {(x^2-4)} {x-2}$.

If we put x = 2 here so it will give us indeterminant of type $\frac {0}{0}$.

So apply limit here:

$\frac {(x^2-4)} {x-2}
= \frac {(x-2)(x+2)} {x-2}
= (x+2) = (2+2) = 4
$

```python
import sympy as sp

x = sp.symbols('x')
expr = (x**2 - 4) / (x - 2)
result = sp.limit(expr, x, 2)
print(result)
```

```python
# two side limit

import sympy as sp

x = sp.symbols('x')
expr = (x**2 - 4) / (x - 2)
left_result = sp.limit(expr, x, 2, dir="-")
right_result = sp.limit(expr, x, 2, dir="+")
print(left_result, right_result)
```

## Derivative

In mathematics, a derivative measures how a function's output changes when its input changes. It tells you the instantaneous rate of change of a quantity or the slope of the tangent line to a curve at a specific point.

### Some common derivatives

**Constant**:
The derivative of Constant is zero; as output value doesn't change with the input value.

$y = f(x) = c$

Slope = $\frac {\Delta y} {\Delta x} = \frac {c - c}{x_1 - x_0} = 0$

```python
import sympy as sp
def const(x):
  return 2

x = sp.symbols('x')
expr = const(x)
result = sp.diff(expr, x)
print(result)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,100)
plt.plot(x,const(x), color="red", label="f(x)");
plt.grid()
plt.axvline(0, color="black")
plt.axhline(0, color="black")
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.legend()
plt.show()
```

**Derivative of a line**

$y = f(x) = ax + b$

Slope = $f'(x) = a$

```python
import sympy as sp

def f(x):
  b = 2
  a = 2
  return a*x + b

x = sp.symbols('x')
expr = f(x)
result = sp.diff(expr, x)
print(result)
```

**Quadratic**

$y = f(x) = x^2$

Slope = $\frac {\Delta f}{\Delta x} = 2x$

```python
import sympy as sp

def f(x):
  return x**2

x = sp.symbols('x')
expr = f(x)
result = sp.diff(expr, x)
print(result)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,100)
plt.plot(x,f(x), color="red", label="f(x)");
plt.grid()
plt.axvline(0, color="black")
plt.axhline(0, color="black")
plt.xlim(-5,5)
plt.ylim(-2,10)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

x_sym = sp.symbols("x")
expr = x_sym**2

y_func = sp.lambdify(x_sym, sp.diff(expr, x_sym), modules="numpy")

x_vals = np.linspace(-10, 10, 100)
y_vals = y_func(x_vals)

plt.plot(x_vals, y_vals, color="red", label="f'(x)=2x")
plt.grid()
plt.axvline(0, color="black")
plt.axhline(0, color="black")
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.legend()
plt.show()
```

## Inverse Function
An inverse function is a function that "undoes" the action of another function.

e.g. f(x) = $3x + 9$

$f^{-1}(x) = \frac {x-9}{3}$


```python
def f(x):
  return 3*x + 9

def f_inv(x):
  return (x-9)/3

print(f(2))
print(f_inv(15))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,100)
plt.plot(x,f(x), color="red", label="f(x)")
plt.plot(x,f_inv(x), color="blue", label="f_inv(x)")
plt.plot(x, x, color="gray", linestyle="--", label="y = x")
plt.grid()
plt.axvline(0, color="black")
plt.axhline(0, color="black")
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.legend()
plt.show()
```

## Differentiable Functions Vs Non-Differentiable Functions

A differentiable function is the one which derivative exists at every point. i,e. at each point there exists a unique tangent line. e.g. $x^2$ is differentiable function.

A Non-Differentiable function is the one which lacks a valid, single-valued derivative at one or more points in its domain. i,e. There exists many tangent lines at one point. e.g. f(x) = |x| is a non-differentiable function.

If a graph has the followong points then it is non-differentiable.
- has corner
- has jump or discontinuity
- has vertical slope

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,100)

fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

ax1.plot(x,x**2, color="red", label="f(x)=x^2")
ax1.axvline(0, color="black")
ax1.axhline(0, color="black")
ax1.set_xlim(-10,10)
ax1.set_ylim(-5,10)
ax1.grid()
ax1.legend()

ax2.plot(x,np.abs(x), color="blue", label="f(x)=|x|")
ax2.axvline(0, color="black")
ax2.axhline(0, color="black")
ax2.set_xlim(-10,10)
ax2.set_ylim(-5,20)
ax2.grid()
ax2.legend();
```

## Rules
- Constant Rule: $\frac {d}{dx}(c) = 0$
- Constant Multiple Rule: $\frac {d}{dx}[cf(x)] = cf'(x)$
- Power Rule: $\frac {d}{dx}(x^n) = nx^{n-1}$
- Sum Rule: $\frac {d}{dx}[f(x)+g(x)] = f'(x)+g'(x)$
- Difference Rule: $\frac {d}{dx}[f(x)-g(x)] = f'(x)-g'(x)$
- Product Rule: $\frac {d}{dx}[f(x)g(x)] = f(x)g'(x)+f'(x)g(x)$
- Quotient Rule: $\frac {d}{dx}[\frac {f(x)}{g(x)}] = \frac {g(x) f'(x) - f(x)g'(x)}{g(x)^2} $
- Chain Rule: $\frac {d}{dx}[f(g(x))] = f'(g(x))g'(x)$

## Differentiation approaches in Python.
- **Symbolic Differentiation:** This approach finds the exact analytical mathematical derivative of an expression, just like you would by hand.
- **Numerical Differentiation:** This approach approximates derivatives using finite differences. It evaluates the function at nearby grid points rather than looking at its algebraic structure.
- **Automatic Differentiation**: Autodiff maps code logic onto a directed acyclic computational graph. It breaks complex software programs down into elementary arithmetic units, then tracks operations using an internal trace ("tape") or source transformation to systematically apply the chain rule.

## Single-Variable Optimization
- Critical Points: Locations where $(f'(x) = 0)$ (flat slope) or $(f'(x))$ is undefined. Candidates for **minima** or **maxima.**
- Second Derivative Test: Evaluates curvature $(f''(x))$ at a critical point:
  -  $(f''(x) > 0)$: Concave up $(\cup ) \rightarrow $ Local Minimum.
	- $(f''(x) < 0)$: Concave down $(\cap ) \rightarrow$ Local Maximum.
	- $(f''(x) = 0)$: Inconclusive (could be an inflection point like $(x^{3})$).


## Partial Derivatives & The Gradient

-  Partial Derivative $(\frac{\partial f}{\partial x_{i}})$: Take the derivative with respect to $(x_{i})$ while treating all other variables as frozen constants.
-  The Gradient $(\nabla f)$: A vector bundling all partial derivatives:
$$\nabla f(x_{1},\dots ,x_{n})=\left[\begin{matrix}\frac{\partial f}{\partial x_{1}}\\ \vdots \\ \frac{\partial f}{\partial x_{n}}\end{matrix}\right]$$


## Gradient Descent

-  Update Rule: Step opposite to the gradient to minimize loss $(L\theta)$:
$$\theta _{t+1}=\theta _{t}-\alpha \nabla _{\theta }L(\theta _{t})$$

- Learning Rate $(\alpha )$: Controls step size.
	- **Too small:** Takes forever to converge; gets stuck easily.
	- **Too big:** Overshoots the valley and oscillates wildly or diverges to $\infty$.


## Multivariable Chain Rule (Backpropagation)

When a loss $L$ depends on intermediate variables $u$ and $v$, which both depend on weight $w$, you sum the derivatives along every path from $L$ to $w$:
$$\frac{\partial L}{\partial w}=\frac{\partial L}{\partial u}\frac{\partial u}{\partial w}+\frac{\partial L}{\partial v}\frac{\partial v}{\partial w}$$

**Backpropagation** is simply the dynamic programming application of this rule: compute gradients at the output layer once, cache them, and multiply backward layer-by-layer so shared sub-expressions are never recomputed.


## Jacobian and Hessian

-  Jacobian (J): Matrix of all first-order partial derivatives for a vector-valued function $\mathbf{f}: \mathbb{R}^n \to \mathbb{R}^m (m \times n$ matrix):

$$J_{ij}=\frac{\partial f_{i}}{\partial x_{j}}$$

- Hessian (H): Square $(n \times n)$ symmetric matrix of all second-order partial derivatives of a scalar function $f: \mathbb{R}^n \to \mathbb{R}$:

$$H_{ij}=\frac{\partial ^{2}f}{\partial x_{i}\partial x_{j}}$$

It captures the curvature of the loss landscape in every direction.
