## Vector
Vector is 1D Array (list that store similar items). It can be either 1D, 2D, 3D or ND in space.

Remember: 1D Array can hold upto nd values. It means that D of 1D array doesn't mean the physical dimension, but it represents the number of indices required to locate the particular element of the list.

```python
import numpy as np

arr_1d = np.array([1])
arr_2d = np.array([1,2])
arr_3d = np.array([1,2,3])

print(arr_1d.ndim, arr_2d.ndim, arr_3d.ndim)
print(arr_1d, arr_2d, arr_3d)
```

As we can clearly see that the number of dimension of array is 1 it means it is a vector but it holds different values.
- `arr_1d` is a 1D array representing a point in 1D space ($x$).
- `arr_2d` is a 1D array representing a point in 2D space ($x,y$).
- `arr_3d` is a 1D array representing a point in 3D space ($x,y,z$).

### Vector Operations

#### 1. **Addition**
Vectors to be added must have the same dimension as vector addition occurs as element-wise.

Mathematically, we have to add the correspoding elements together. e.g. $u = [1,2]$ and $v = [3,4]$ so the result will be $r = u+v = [1+3, 2+4] = [4,6]$

Geometically you can represent vectors according to **Head-to-Tail** rule (we put the second vector tail on the first vector head). In addition, it doesn't matter if you put the first one tail on the second or second one tail on the first the result will be the same (commutative property holds here).
The result will be a straight line drawn from the tail of the first vector to the head of the last vector.

$$\mathbf{u} + \mathbf{v} = \begin{bmatrix} u_1 \\ u_2 \end{bmatrix} + \begin{bmatrix} v_1 \\ v_2 \end{bmatrix} = \begin{bmatrix} u_1 + v_1 \\ u_2 + v_2 \end{bmatrix}$$

```python
import numpy as np
import matplotlib.pyplot as plt

u = np.array([1,3])
v = np.array([4,2])

plt.figure(figsize=(10,4))
plt.quiver(0,0,u[0],u[1], color="blue", angles='xy', scale_units='xy', scale=1)
plt.quiver(0,0,v[0],v[1], color="red", angles='xy', scale_units='xy', scale=1)
plt.xlim(-1,8)
plt.ylim(-1,8)
plt.grid(True, alpha=0.5)
plt.axhline(0, color='gray')
plt.axvline(0, color='gray')
plt.title("Representation of vector u and v")
plt.show()
```

```python
print("Addition: ",v+u)

plt.figure(figsize=(10,4))
plt.quiver(0,0,u[0],u[1], color="blue", angles='xy', scale_units='xy', scale=1, label="u")
plt.quiver(u[0],u[1],v[0],v[1],color="red", angles='xy', scale_units='xy', scale=1, label="v")
plt.quiver(0,0,v[0],v[1], color="red", angles='xy', scale_units='xy', scale=1)
plt.quiver(v[0],v[1],u[0],u[1],color="blue", angles='xy', scale_units='xy', scale=1)
plt.quiver(0,0,(u+v)[0],(u+v)[1], color="orange", angles='xy', scale_units='xy', scale=1, label="u+v")
plt.legend()
plt.xlim(-1,8)
plt.ylim(-1,8)
plt.grid(True, alpha=0.5)
plt.axhline(0, color='gray')
plt.axvline(0, color='gray')
plt.show()
```

Now the question arise that how to add more than two vectors? The answer is just add their corresponding elements together. e.g. u+v+w = [x1+x2+x3 , y1+y2+y3]. Mathematically, just put the tail of one over the head of another and move on up to last vector. Then just draw a vector from origin to the last vector head and we have our result.

```python
u = np.array([1,3])
v = np.array([4,2])
w = np.array([2,1])

print(u+v+w)
```

```python
u = np.array([1,3])
v = np.array([4,2])
w = np.array([2,2])
x = np.array([1,2])
y = np.array([2,2])
z = np.array([3,1])

plt.quiver(0,0,u[0],u[1],angles='xy',scale_units='xy',scale=1, color="blue", label="u")
plt.quiver(u[0],u[1],v[0],v[1],angles='xy',scale_units='xy',scale=1, color="red", label="v")
plt.quiver((u+v)[0],(u+v)[1],w[0],w[1],angles='xy',scale_units='xy',scale=1, color="green", label="w")
plt.quiver((u+v+w)[0],(u+v+w)[1],x[0],x[1],angles='xy',scale_units='xy',scale=1, color="magenta", label="x")
plt.quiver((u+v+w+x)[0],(u+v+w+x)[1],y[0],y[1],angles='xy',scale_units='xy',scale=1, color="pink", label="y")
plt.quiver((u+v+w+x+y)[0],(u+v+w+x+y)[1],z[0],z[1],angles='xy',scale_units='xy',scale=1, color="cyan", label="z")

plt.quiver(0,0, (u+v+w+x+y+z)[0], (u+v+w+x+y+z)[1], angles='xy', scale_units='xy', scale=1, color="black", label="R")
plt.grid(True, alpha=0.5)
plt.axhline(0, color='gray')
plt.axvline(0, color='gray')
plt.legend()
plt.xlim(-1,16)
plt.ylim(-1,14)
plt.show()
```

Now how can we add vectors representing 3D space? Again we will add their corresponding elements.
e.g. u+v = [x1+x2, y1+y2, z1+z2] and geometrically we can apply the same **Head-to-Tail**  rule.

#### 2. **Subtraction**

For subtraction of vector, we must have the same dimension of both vectors.

Mathematically, we subtract the correspoding components. e.g. $u = [1,2]$ and $v = [3,4]$ so the result will be $r = u-v= [1-3, 2-4] = [-2,-2]$

Geometically, we can apply two methods.

- Method 1: Draw vector A and vector B from the same origin. Then draw a vector from the Head of B to the head of A which will be the resultant vector (the arrow always points to vector we are subtracting from (first vector)).
- Method 2: Flip the arrow B and put it on the head of arrow A and then draw from a resultant vector from the arrow A towards the head of arrow B.





```python
import numpy as np

u = np.array([5,2])
v = np.array([2,4])

u-v
```

```python
plt.quiver(0,0,u[0],u[1],angles='xy',scale_units='xy',scale=1, color="blue", label="u")
plt.quiver(0,0,v[0],v[1],angles='xy',scale_units='xy',scale=1, color="red", label="v")
plt.quiver(v[0],v[1],(u-v)[0],(u-v)[1],angles='xy',scale_units='xy',scale=1, color="orange", label="u-v")
plt.legend()
plt.xlim(-1,8)
plt.ylim(-1,8)
plt.grid(True, alpha=0.5)
plt.axhline(0, color='gray')
plt.axvline(0, color='gray')
plt.show()
```

```python
plt.quiver(0,0,u[0],u[1],angles='xy',scale_units='xy',scale=1, color="blue", label="u")
plt.quiver(u[0],u[1],v[0],v[1],alpha=0.5, angles='xy',scale_units='xy',scale=1, color="red", label="v")
plt.quiver(u[0],u[1],-v[0],-v[1],angles='xy',scale_units='xy',scale=1, color="red", label="-v")
plt.quiver(0,0,(u-v)[0],(u-v)[1],angles='xy',scale_units='xy',scale=1, color="green", label="u+(-v)")
plt.legend()
plt.xlim(-1,10)
plt.ylim(-5,8)
plt.grid(True, alpha=0.5)
plt.axhline(0, color='gray')
plt.axvline(0, color='gray')
plt.show()
```

#### 3. **Scalar Multiplication**

Scalar when multiplied with a vector can stretch, shrink or flip its direction.

Mathematically, we just multiply that scalar with each component of the vector. e.g. 3 * [2,1] = [6,3]

Geometrically, scalar can either stretch it or shrink it. Negative scalar can flip its direction.

$$c \cdot \mathbf{v} = c \cdot \begin{bmatrix} v_1 \\ v_2 \end{bmatrix} = \begin{bmatrix} c \cdot v_1 \\ c \cdot v_2 \end{bmatrix}$$

```python
import numpy as np

a = np.array([3,4])
positive_k = 2
negative_k = -2
zero_k = 0

print("Positive Scalar: ", positive_k * a)
print("Negative Scalar: ", negative_k * a)
print("Zero Scalar: ", zero_k * a)
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.quiver(0,0,*a, angles='xy', scale_units='xy', scale=1, color="black", label="a")
plt.quiver(0,0,*(positive_k * a), angles='xy', scale_units='xy', scale=1, color="blue", label="2a",alpha=0.2)
plt.quiver(0,0,*(negative_k * a), angles='xy', scale_units='xy', scale=1, color="red", label="-2a")
plt.quiver(0,0,*(zero_k * a), angles='xy', scale_units='xy', scale=1, color="green", label="0a")

plt.grid(True, alpha=0.5)
plt.axhline(0, color='gray')
plt.axvline(0, color='gray')
plt.legend()
plt.xlim(-10,10)
plt.ylim(-10,10)
```

#### 4. **Hadamard Product**

It is also known as Element-wise Product. In it two vectors of the same size can be multiplied together element-wise.

Symbolically, it can be represented as $\odot$.

let a = [1,2], and b = [2,3] so $a \odot b$ = [1 * 2, 2 * 3] = [2,6]

Note: Geometrically, there is no single, fixed geometric arrow rule for the Hadamard product, but it represents scaling up the vector.

$$\mathbf{u} \odot \mathbf{v} = \begin{bmatrix} u_1 \cdot v_1 \\ u_2 \cdot v_2 \end{bmatrix}$$

#### 6. Dot Product

Dot Product take two equal length vectors and multiply them (not element-wise) and give us back a single scalar value.

Mathematically: Imagine we have a = [x1, y1] and b = [x2, y2] and we have to calculate dot product of a and b so a.b = x1*x2+y1*y2 = k where k is a single scalar value. For a vector having three values it can be represented as a.b = x1*y1+x2y2+x3y3 = k.

Now what does it show geometrically?
Geometrically, it shows the shadow of one vector casts on another one and tells that whether they are pointing the same direction, opposite or perpendicular to each other. Dot product is just a value when we multiply the projection of vector A with the length of vector B (projection of A onto b * length of vector B).

Now,
- If dot product is positive then the two vectors are work together and in the same direction.
- If dot product is negative then two vectors are in opposite direction and they are fighting each other.
- If dot product is zero then both vectors don't have affect on each other.

But now you might ask that why we are using the dot product for the same function that the angle do? The answer is the dot product give magnitude alongside with the direction. moreover the computation of angle is quite complex (require more computational power) than the dot product.

Note: The dot product (scalar) depends on the length of vectors and their alignment both.

Algebraic Formula:$$\mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^n u_i v_i = u_1 v_1 + u_2 v_2 + \dots + u_n v_n$$
Geometric Formula:$$\mathbf{u} \cdot \mathbf{v} = \Vert{}\mathbf{u}\Vert{}_2 \Vert{}\mathbf{v}\Vert{}_2 \cos(\theta)$$

We will discuss this geometric fomrula a little later.

Here something that we must note that when we plot two vectors from the same origin and find the angle between them will give us the following information:
- In case of acute angle (< 90), the two vector works together.
- In case of right angle (90), the two vector have no affect on each other.
- In case of obtuse angle (> 90), the two vectors works in the opposite direction.


```python
import numpy as np
a = np.array([2,4])
b = np.array([6,3])

print(np.dot(a,b))
# It can also be represented as a @ b
print(a@b)
```

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([6, 2])
b = np.array([2, 5])

dot_product = np.dot(a, b)
print(f"Dot Product (a · b): {dot_product}")
projection_vector = (dot_product / np.dot(a, a)) * a

plt.figure(figsize=(8, 6))
max_val = max(np.max(a), np.max(b)) + 2
plt.quiver(0, 0, a[0], a[1], color="red", angles='xy', scale_units='xy', scale=1, label=f'Vector a {a}')
plt.quiver(0, 0, b[0], b[1], color="blue", angles='xy', scale_units='xy', scale=1, label=f'Vector b {b}')
plt.quiver(0, 0, projection_vector[0], projection_vector[1], color="green",
           angles='xy', scale_units='xy', scale=1, label='Projection of b onto a', alpha=0.7)
plt.plot([b[0], projection_vector[0]], [b[1], projection_vector[1]], 'k--', alpha=0.6, label='Perpendicular Drop')

plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.grid(True, alpha=0.3)
plt.legend()
plt.xlim(-1, max_val)
plt.ylim(-1, max_val)
plt.title(f"Geometric Dot Product Visualization\n(a · b = {dot_product})")
plt.gca().set_aspect('equal')
plt.show()
```

#### 7. Cross Product

The cross product multiply two vectors and give us the brand new vector as a result. It always works in 3D.

Let's suppose we have  $\mathbf{u} = (u_1, u_2, u_3)$ and $\mathbf{v} = (v_1, v_2, v_3)$

Now to calculate the cross product first conver this into 3x3 matrix.
$$\mathbf{u}\times \mathbf{v}=\left|\begin{matrix}\mathbf{i}&\mathbf{j}&\mathbf{k}\\ u_{1}&u_{2}&u_{3}\\ v_{1}&v_{2}&v_{3}\end{matrix}\right|$$
Now find the determinant

$$\mathbf{u}\times \mathbf{v}=\mathbf{i}(u_{2}v_{3}-u_{3}v_{2})-\mathbf{j}(u_{1}v_{3}-u_{3}v_{1})+\mathbf{k}(u_{1}v_{2}-u_{2}v_{1})$$

simplify to vector form

$$(u_{2}v_{3}-u_{3}v_{2},\quad -(u_{1}v_{3}-u_{3}v_{1}),\quad u_{1}v_{2}-u_{2}v_{1})$$

Now what geometrically it represents?
When we perform cross product of two vectors then we get the new vector perpendicular to both vectors. The magnitude of the new vector is equal to the area of parallelogram formed by the already existing vectors.
- If vectors points the same direction so no area is present so cross product is zero.
- If vectors form right angle so area is maximum so the we get the longest vector.

To find the direction we can use **The Right-Hand Rule**. Point your index finger towards the first vector, point the middle towards the second vector you thumb will show the direction of the third vector.

$$|{}|{}\vec{a}\times \vec{b}|{}|{}=|{}|{}\vec{a}|{}|{}\cdot |{}|{}\vec{b}|{}|{}\cdot \sin (\theta )$$

We will cover this formula later on.

```python
import numpy as np


u = np.array([3, 1, 0])
v = np.array([0, 3, 1])

print("Result: ",np.cross(u, v))
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

u = np.array([3, 0, 0])
v = np.array([0, 3, 0])

p0 = np.array([0, 0, 0])
p1 = u
p2 = u + v
p3 = v

vertices = [[p0, p1, p2, p3]]
parallelogram = Poly3DCollection(vertices, facecolor='orange', edgecolor='darkorange', alpha=0.3)
ax.add_collection3d(parallelogram)

ax.plot3D([0, u[0]], [0, u[1]], [0, u[2]], 'red', label='u')
ax.plot3D([0, v[0]], [0, v[1]], [0, v[2]], 'blue', label='v')
ax.plot3D([0, np.cross(u, v)[0]], [0, np.cross(u, v)[1]], [0, np.cross(u, v)[2]], 'green', label='u x v')

plt.legend()
ax.set_title("Cross Product Visualization")
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
plt.show()
```

#### Norm of a vector

Norm of a vector represents the length or size of the vector.

You might be asking that why do we need norm, i mean we have coordinates so is it not enough? The coordinate represent the location in space while the actual distance is represented by norm like how much far the vector is from the origin. In graph having many points, it is difficult that you tell the which vector is bigger than other (in length). so for that we used norms.


Now to find the distance from the origin to the head of a vector we have many ways and thus we can say we have different types of norms.

- **$L_{1}$ Norm (Manhattan / Taxicab Norm)**:
 The sum of the absolute values of the coordinates. d = |x|+|y|
- **$L_{2}$ Norm (Euclidean Norm)**:
 The standard straight-line geometric length. d = $\sqrt{x^2+y^2}$
 - **$L_{inf}$ Norm**:
  It is also known as maximum norm or uniform norm, measure the magnitude of a vector's component having highest value. d = $max |x, y|$.
  It only depends on the largest component. It measure the single longest directional move.
- **L_{p} Norm**: We take each number, raise it to the power of p, add them all together, and then take the p-th root of the total.
 As p changes, the result changes:
  - If p=1: $((3^1 + 4^1)^1 = \mathbf{7})$ (The $(L_{1})$ norm)
  - If p=2: $(\sqrt{3^2 + 4^2} = \mathbf{5})$ (The $(L_{2}$) norm)
  - If p=$\infty$: It structurally morphs into $(max(3, 4) = \mathbf{4})$ (The $(L_{\infty })$ norm)
  


```python
import numpy as np
v = np.array([3,4])
np.linalg.norm(v, ord=1) # L1 Norm
```

### Cosine Similarity

Cosine Similarity is used to measure how similar two vectors are, regardless of their size.

- If result is +, so the vectors are in the same direction.
- If result is -, so the vectors are in the opposite direction.
- If result is 0, so both vectors are perpendicular to each other.

Now you might ask that why do we need cosine similarity as we can get the same idea from the angle between them? The answer lies in the mathematics of it.

$$\text{Cosine Similarity}=\cos (\theta )=\frac{\mathbf{A}\cdot \mathbf{B}}{\|{}\mathbf{A}\|{}\times \|{}\mathbf{B}\|{}}$$

To find angle, you must take arccos which is too computational.


```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([4,0])
b = np.array([3,3])

dot_product = np.dot(a, b)
magnitude_a = np.linalg.norm(a)
magnitude_b = np.linalg.norm(b)

cosine_similarity = dot_product / (magnitude_a * magnitude_b)
cosine_similarity
```

### Types of Matrix

Matrices are categorized based on their shape, pattern of elements or mathematical properties.


#### Based on Shape

1. **Square Matrix**: Same number of rows and columns.
2. **Rectangular Matrix**: Different number of rows and columns.
3. **Row Matrix**: Has only one row. It is also known as Row Vector.
4. **Column Matrix**: Has only one column. It is also known as Column Vector.

#### Based on Element-Pattern

1. **Zero (null) Matrix**: All elements are zero.
2. **Diagonal Matrix**: Have non-zero elements on diagonal while others are zero.
3. **Identity Matrix**: Have 1 on the main diagonal and zeros on other places.
4. **Scalar Matrix**: A square diagonal Matrix where all the main diagonal elements are same number. The rectangular matrix can't be a scalar matrix.
5. **Triangular Matrix**:
  - **Upper Triangular Matrix**: All entries below the main diagonal are 0.
  - **Lower Trainular Matrix**: All entries above the main diagonal are 0.


```python
import numpy as np

zero  = np.zeros((2,3))
diagonal = np.diag([1,2,3])
identity = np.eye(3)
scalar = np.diag([1,1,1])
upper_triangular = np.triu([[1,2,3],[4,5,6],[7,8,9]])
lower_triangular = np.tril([[1,2,3],[4,5,6],[7,8,9]])

print("Zero Matrix: \n",zero)
print("Diagonal Matrix: \n",diagonal)
print("Identity Matrix: \n",identity)
print("Scalar Matrix: \n",scalar)
print("Upper Triangular Matrix: \n",upper_triangular)
print("Lower Triangular Matrix: \n",lower_triangular)
```

#### Based on Property
1. **Symmetric Matrix**: $A = A^T$
2. **Orthogonal Matrix**: $AA^T = I$
3. **Invertible Matrix (Non-Singular Matrix)**: $A^{-1}A = I$
4. **Sparce Matrix**: Most entries are zero.
5. **Singular Matrix**: Has determinant of 0.
6. **Non-Singular Matrix**: Has determinant not equal to 0.

```python
import numpy as np

symmetric_matrix = np.array([[4,1],[1,-2]])
orthogonal_matrix = np.array([[1,0],[0,-1]])
Invertible_matrix = np.array([[1,2],[2,5]])
sparse_matrix = np.array([[1,0,0],[0,2,0],[0,0,3]])
singular_matrix = np.array([[1,2],[2,4]])
non_singular_matrix = np.array([[1,2],[2,3]])

print("Symmetric Matrix: \n",symmetric_matrix, "\n", symmetric_matrix == symmetric_matrix.T)
print("-"*10)
print("Orthogonal Matrix: \n",orthogonal_matrix,"\n", orthogonal_matrix @ orthogonal_matrix.T == np.eye(2))
print("-"*10)
print("Invertible Matrix: \n",Invertible_matrix, "\n", np.linalg.inv(Invertible_matrix)@Invertible_matrix == np.eye(2))
print("-"*10)
print("Sparse Matrix: \n",sparse_matrix)
print("-"*10)
print("Singular Matrix: \n",singular_matrix,"\n",np.linalg.det(singular_matrix))
print("-"*10)
print("Non-Singular Matrix: \n",non_singular_matrix,"\n",np.linalg.det(non_singular_matrix))

```

### Matrix as linear Transformation

Linear transformation is a function that take a vector from one vector space and maps it into another vector space while keeping vectors addition and scalar multiplication unchanged.

The function is linear transformation if it satisfy the following conditions:
- Additivity (Superposition): $T(u+v) = T(u) + T(v)$
- Homogeneity (Scaling): $T(c.v) = c.T(v)$

Geometrically, linear transformation has the following properties:
- Origin remains fixed.
- Straight lines stay straight: Grid lines remain straight and evenly spaced.

e.g. Rotation, Scaling, and reflextion are some example of linear transformation.

Matrix are used for unique linear transformation. Each matrix column represents the basis vector of a new space.


```python
import numpy as np
import matplotlib.pyplot as plt

# Standard Basis
i_hat = np.array([1, 0])
j_hat = np.array([0, 1])
v = np.array([1, 1])  # Our custom vector (1*i_hat + 1*j_hat)

# Matrix as linear transformation
A = np.array([[1, 0],
              [0, -1]])

# Apply the transformation using matrix multiplication (@)
i_hat_new = A @ i_hat  # Lands at [1, 0]
j_hat_new = A @ j_hat  # Lands at [0, -1]
v_new = A @ v          # Lands at [1, -1]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# --- Original Standard Basis Space ---
ax1.quiver(0, 0, i_hat[0], i_hat[1], angles='xy', scale_units='xy', scale=1, color='red', label=r'$\hat{i}$ [1, 0]')
ax1.quiver(0, 0, j_hat[0], j_hat[1], angles='xy', scale_units='xy', scale=1, color='green', label=r'$\hat{j}$ [0, 1]')
ax1.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='blue', label=r'$\vec{v}$ [1, 1]')
ax1.set_title("Original Standard Basis Space")

# --- Transformed New Basis Space ---
ax2.quiver(0, 0, i_hat_new[0], i_hat_new[1], angles='xy', scale_units='xy', scale=1, color='red', label=r'New $\hat{i}$ [1, 0]')
ax2.quiver(0, 0, j_hat_new[0], j_hat_new[1], angles='xy', scale_units='xy', scale=1, color='darkgreen', label=r'New $\hat{j}$ [0, -1]')
ax2.quiver(0, 0, v_new[0], v_new[1], angles='xy', scale_units='xy', scale=1, color='purple', label=r'Transformed $\vec{v}$ [1, -1]')
ax2.set_title("Transformed Space (Reflected over X-axis)")

for ax in [ax1, ax2]:
    ax.set_xlim(-2, 2)
    ax.set_ylim(-2, 2)
    ax.grid(True, linestyle='--')
    ax.axhline(0, color='black', lw=0.5)
    ax.axvline(0, color='black', lw=0.5)
    ax.set_aspect('equal')
    ax.legend()

plt.tight_layout()
plt.show()
```

Now, we have seen that matrix transformed the whole space and now we have new vector space where the old vector as represented by its new coordinates.

Now lets see how matrix transformed the overall space.

```python
import numpy as np
import matplotlib.pyplot as plt


# This matrix will slide the grid horizontally and scale it
A = np.array([
    [1.5, 1.0],  # Transforms x
    [0.0, 1.0]   # Transforms y
])

def T(v):
    """Applies the linear transformation T(v) = A @ v"""
    return A @ v

print("--- PART 1: ANALYTICAL VERIFICATION ---")

u = np.array([2.0, 3.0])
v = np.array([-1.0, 4.0])
c = 2.5

lhs_add = T(u + v)
rhs_add = T(u) + T(v)
additivity_passed = np.allclose(lhs_add, rhs_add)
print(f"1. Additivity Test T(u+v) == T(u)+T(v): {additivity_passed}")
print(f"   T(u+v)   = {lhs_add}")
print(f"   T(u)+T(v) = {rhs_add}\n")

lhs_hom = T(c * v)
rhs_hom = c * T(v)
homogeneity_passed = np.allclose(lhs_hom, rhs_hom)
print(f"2. Homogeneity Test T(c*v) == c*T(v): {homogeneity_passed}")
print(f"   T(c*v)   = {lhs_hom}")
print(f"   c*T(v)   = {rhs_hom}\n")

zero_vector = np.array([0.0, 0.0])
origin_passed = np.allclose(T(zero_vector), zero_vector)
print(f"3. Origin Test T(0) == 0: {origin_passed}")
print(f"   T(0)     = {T(zero_vector)}")



x = np.linspace(-4, 4, 9)
y = np.linspace(-4, 4, 9)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

# Before transformation
for xi in x:
    ax1.plot([xi, xi], [y[0], y[-1]], color='blue', alpha=0.3, linestyle='--')
for yi in y:
    ax1.plot([x[0], x[-1]], [yi, yi], color='blue', alpha=0.3, linestyle='--')
ax1.scatter(0, 0, color='red', s=100, zorder=5, label='Origin (0,0)')
ax1.set_title("Original Grid\n(Straight, Orthogonal, Fixed Origin)")
ax1.set_xlim(-6, 6)
ax1.set_ylim(-6, 6)
ax1.grid(True, which='both', alpha=0.1)
ax1.axhline(0, color='black',linewidth=1)
ax1.axvline(0, color='black',linewidth=1)
ax1.legend()

# After Transformation
for xi in x:
    p1 = np.array([xi, y[0]])
    p2 = np.array([xi, y[-1]])
    p1_trans = T(p1)
    p2_trans = T(p2)
    ax2.plot([p1_trans[0], p2_trans[0]], [p1_trans[1], p2_trans[1]], color='purple', alpha=0.7)

for yi in y:
    p1 = np.array([x[0], yi])
    p2 = np.array([x[-1], yi])
    p1_trans = T(p1)
    p2_trans = T(p2)
    ax2.plot([p1_trans[0], p2_trans[0]], [p1_trans[1], p2_trans[1]], color='purple', alpha=0.7)

origin_trans = T(np.array([0, 0]))
ax2.scatter(origin_trans[0], origin_trans[1], color='red', s=100, zorder=5, label='Transformed Origin')

ax2.set_title("Transformed Grid T(v) = Av\n(Lines stay straight & evenly spaced, Origin fixed)")
ax2.set_xlim(-10, 10)
ax2.set_ylim(-6, 6)
ax2.grid(True, which='both', alpha=0.1)
ax2.axhline(0, color='black',linewidth=1)
ax2.axvline(0, color='black',linewidth=1)
ax2.legend()

plt.tight_layout()
plt.show()
```

## System of Linear Equation

A system of linear equations is simply a collection of two or more linear equations that share the same set of unknown variables.

To solve system of linear equation, we must be sure that system must not have redundunt information and neither have contradictory information, but have complete information.
We can solve the system of linear equation in two ways:
- Analytical
- Geometrical

### Analytical Method

Imagine: 1 apple + 1 banana = 12
 1 apple + 2 banana = 14
 apple cost, banana cost = ?

 If we use our common sense then adding one banana only increase the price by $2 so it means that banana price is 2 while the apple price is 10

 let's check it analytically

 a + b = 12 -> 1
 a + 2b = 14 -> 2

subtract eq 1 from eq 2:
$(a + 2b) - (a + b) = 14 - 12$
b = 2

put the value of b into eq 1:
a + 2 = 12
a = 10

Note to solve this in python we must convert this into matrix form:

$$ A =
\begin{bmatrix}
1 & 1 \\
1 & 2
\end{bmatrix}
$$

$$ x =
\begin{bmatrix}
x \\
y
\end{bmatrix}
$$

$$ b =
\begin{bmatrix}
12 \\
14
\end{bmatrix}
$$

so $$Ax = b$$


```python
import numpy as np

A = np.array([[1, 1], # 1 apple + 1 banana
              [1, 2]]) # 1 apple + 2 banana
B = np.array([12, 14]) # price: 12 , 14

solution = np.linalg.solve(A, B)
a, b = solution
print(f"a = {a}")
print(f"b = {b}")
```

## Gaussian Elemination

To solve the system of linear equation which form the complex matrix form, we use gaussian elemination method.

This method converts the system into augmented matrix (matrix of equation with the answer (usually separated by |)) and then convert this into upper triangular matrix or row echelon form.

Row echelon form is the form of the matrix which follows the following conditions:
- The rows consists of zero must be at the bottom.
- The first non-zero element (pivot) must be at the right of the below row's pivot.
- All the element below the pivot must be zero.

Gaussian elemination allowed the following operations:
- swapping: switch the position of two rows.
- scaling: multipy or divide the entire row by scalar.
- add/subtract: add or subtract two rows.

The Process of Gaussian Elimination:
- Convert the system of linear equation into augmented matrix.
- Use operations to form row echelon form.
- Back substitute the values so to get the answer.

Row Echelon form can be represented as:
$$\left[\begin{array}{ccc|c}\mathbf{2}&4&-2&2\\ 0&\mathbf{-3}&6&9\\ 0&0&\mathbf{5}&10\end{array}\right]$$

where 5z = 10, -3y + 6z = 9, and 2x + 4y -2z =2.

Reduced row echelon form is the one following these condition:
- must have row echelon form.
- all the pivot must be 1.
- above and below of the pivot must be zero.

$$\left[\begin{array}{ccc|c}\mathbf{1}&0&0&1\\ 0&\mathbf{1}&0&-1\\ 0&0&\mathbf{1}&2\end{array}\right]$$

where x = 1, y = -1, z = 2.

Note: Gaussian Elimination give us row echelon form while Gauss-Jordan Elimination give us reduced row echelon form.

### Rank of Matrix

The rank of a matrix represents the number of truly unique, non-redundant pieces of information contained within that matrix.

To find rank we convert the matrix into row echelon form or reduced row echelon form and then the number of rows having a pivot (number of non-zero rows) represents a rank.

```python
import numpy as np

A = np.array([[1, 2, 3],
              [0, 1, 5]])

rank = np.linalg.matrix_rank(A)
print(rank)
```

### Some important topics

Before jumping into Matrix Inverse, we have to know about some import concepts.

**EigenVector and EigenValue**: When we apply matrix transformation on vectors, then most of the vectors changed their position while there are some special vectors that remains on its place and just scaled up or down. Those vectors are called Eigenvectors and by the amount they scaled up or down is called Eigenvalues.

The formula for this is $$Av = \lambda v$$ where A = matrix, v = vector, λ = eigenvalue.

To find eigenvalue:
$$Av = \lambda Iv$$
$$Av - \lambda Iv = 0$$
$$(A-\lambda I)v = 0$$
Here either v = 0 or the matrix. By definition v can't be zero so A-λ I = 0.
$$det(A - \lambda v) = 0$$

To find the eigenvector, we put the value of lamnbda in in $det(A -\lambda v) = 0$ and then solve it to get the eigenvector.

```python
import numpy as np

# 1. Define the square matrix A
A = np.array([[4, 1],
              [2, 3]])

# 2. Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

# 3. Print the results
print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors (as columns):")
print(eigenvectors) # numpy give output of normalized vector.
```

**Spectral Decomposition**: It is also known as orthogonal decomposition.
If Matrix A has linear indpenedent eigenvectors so it can be written as:
$$A = XDX^{-1}$$
Where,
- A is our original square matrix.
- X is a matrix whose columns are the eigenvectors of A.
- Dis a diagonal matrix whose diagonal entries are the corresponding eigenvalues of A.
- $X^{-1}$ is the inverse of the eigenvector matrix.

If original matrix is symmetric, then the maths become easy. As we know that symmetric matrix has special property $A^{-1} = A^T$ so the above equation becomes:
$$A = QDQ^{T}$$
(Where Q is the matrix of normalized, perpendicular eigenvectors).

We know that Q and $Q^{T}$ are orthogonal matrix which perform rotation, while D is a diagonal matrix which do scaling.

```python
import numpy as np

A = np.array([[4, 2], [2, 4]])

#  Compute eigenvalues (Lambda) and orthogonal eigenvectors (Q)
eigenvalues, Q = np.linalg.eigh(A)

# Form the diagonal matrix of eigenvalues
Lambda = np.diag(eigenvalues)

# Reconstruct A using spectral decomposition: A = Q @ Lambda @ Q^T
A_reconstructed = Q @ Lambda @ Q.T

print("Original Matrix A:\n", A)
print("\nEigenvalues:\n", eigenvalues)
print("\nEigenvector Matrix Q:\n", Q)
print("\nReconstructed A (Q * Lambda * Q^T):\n", A_reconstructed)
```

**Singular Value Decomposition (SVD)**: A matrix can be written as:
$$A = U \Sigma V^T$$

where U and V are orthogonal matrices and Σ is a diagonal matrix containing non-negative singular values.

Steps required for this:
- First computer $A^TA$
- Find the eigen values and then arange it in descending order.
- Take squre root of each value so we have singular values. make matrix of it which becomes our diagonal matrix $\Sigma$.
- Find the Eigenvectors of $A^{T}A$ to construct V.
- Repeat the same process for $AA^T$ to construct U.


```python
import numpy as np

A = np.array([[1, 2],
              [3, 4],
              [5, 6]])


U, S, VT = np.linalg.svd(A, full_matrices=False)

print("U matrix:\n", U)
print("\nSingular values (S):", S)
print("\nVT matrix:\n", VT)

Sigma = np.diag(S)
A_reconstructed = U @ Sigma @ VT

print("\nReconstructed Matrix A:\n", A_reconstructed)
```

### Matrix inverse

Matrix Inverse can be represented as $A^{-1}$

$AA^{-1} = I$

The Square matrix inversion can be calculated as:

- **Computing via Gauss-Jordan Elimination**: you write the matrix in this form $[A|I]$ and then by using row operations you make it like $[I|A^{-1}]$.
- **Computing via the Adjugate Formula**:
$\mathbf{A}^{-1}=\frac{1}{\det (\mathbf{A})}\text{adj}(\mathbf{A})$.

But now you might think how can we find the matrix inversion of non-square (rectangular) matrix?

**More-Penrose pseudo-inverse**: The most common method used for matrix inverse for non-square or singular matrices.
The formula for finding the Moore-Penrose pseudoinverse ($A^{+}$) using the components of your SVD $(A = U \Sigma V^T)$ is:

$$A^{+}=V\Sigma ^{+}U^{T}$$

To find $\Sigma ^{+}$, you perform two simple operations on the diagonal matrix ($\Sigma $):
1. Invert all non-zero singular values (replace $\sigma _{i}$ with $\frac{1}{\sigma _{i}}$).
2. Leave any zeros exactly as 0.
3. Transpose the matrix (if it's non-square, its dimensions flip)

```python
import numpy as np

A = np.array([[1, 2],
              [2, 1]])

# Find the Moore-Penrose Pseudoinverse (Works for ANY matrix)
A_pseudoinverse = np.linalg.pinv(A)

# Find the Standard Inverse (Only works for square, invertible matrices)
A_standard_inverse = np.linalg.inv(A)

print("Moore-Penrose Pseudoinverse:\n", A_pseudoinverse)
print("\nStandard Inverse:\n", A_standard_inverse)
```

## Principle Component Analysis (PCA)
Principal Component Analysis (PCA) is a dimensionality reduction technique that brings all these concepts together.

Imagine you have a dataset with many features (e.g., 50 columns in a matrix). Visualizing or modeling 50 dimensions is hard, computationally expensive, and often redundant due to multi-collinearity. PCA finds a new set of perpendicular coordinate axes (called Principal Components) that point in the directions where your data varies the most.

- First Principal Component ($PC_1$): The line/axis through your data that captures the maximum possible variance (information).
- Second Principal Component ($PC_2$): The line perpendicular (orthogonal) to $PC_1$ that captures the maximum remaining variance.
- $PC_k$: Continues this pattern until all dimensions are captured.

By keeping only the top $k$ principal components, you reduce the dimension of your dataset while retaining most of its original information (variance).

### How PCA Works (Step-by-Step Mathematical Intuition)
Since we already know how to compute covariance matrices, eigenvectors, and SVD, here is how the algorithm works mathematically in Python/NumPy terms:

**Step 1: Center the Data (Mean Centering):**

Subtract the mean of each feature (column) from the dataset. This shifts the center of your data cloud directly to the origin $(0,0,\dots,0)$.$$\mathbf{X}_{\text{centered}} = \mathbf{X} - \boldsymbol{\mu}$$

**Step 2: Compute the Covariance Matrix**

Compute how each variable co-varies with every other variable:$$\mathbf{\Sigma} = \frac{1}{n-1} \mathbf{X}_{\text{centered}}^T \mathbf{X}_{\text{centered}}$$

**Step 3: Compute Eigenvalues & Eigenvectors**

Find the eigenvectors and eigenvalues of the covariance matrix $\mathbf{\Sigma}$:$$\mathbf{\Sigma} \mathbf{v}_i = \lambda_i \mathbf{v}_i$$
- Eigenvectors ($\mathbf{v}_i$): Define the directions of the new axes (Principal Components).
- Eigenvalues ($\lambda_i$): Represent the amount of variance captured along each direction.

**Step 4: Sort Components by Variance**

Sort the eigenvalues in descending order ($\lambda_1 \ge \lambda_2 \ge \dots$) and re-order the corresponding eigenvectors.

**Step 5: Project Data onto New Feature Space**

Select the top $k$ eigenvectors to form a feature matrix $\mathbf{W}_k$. Multiply your original centered matrix by $\mathbf{W}_k$ to get the reduced dataset:$$\mathbf{X}_{\text{reduced}} = \mathbf{X}_{\text{centered}} \mathbf{W}_k$$

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate correlated 2D toy data
np.random.seed(42)
x = np.random.normal(5, 2, 200)
y = 0.8 * x + np.random.normal(0, 1, 200)
X = np.vstack((x, y)).T

# 2. Step 1: Mean Center the Data
X_centered = X - np.mean(X, axis=0)

# 3. Step 2: Covariance Matrix
cov_matrix = np.cov(X_centered, rowvar=False)

# 4. Step 3 & 4: Eigendecomposition & Sorting
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Sort descending
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# 5. Step 5: Project onto 1D (Top 1 Principal Component)
W2 = eigenvectors[:, :1]  # Keep 1 component
X_projected_1d = X_centered @ W2

print("Eigenvalues (Variance per Component):", eigenvalues)
print("Explained Variance Ratio:", eigenvalues / np.sum(eigenvalues))
print("Principal Component Direction Matrix:\n", eigenvectors)

# --- Visualization in your notebook style ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Original Data with Principal Component Axes
ax1.scatter(X_centered[:, 0], X_centered[:, 1], alpha=0.4, color='blue', label='Data Points')
for i, (val, vec) in enumerate(zip(eigenvalues, eigenvectors.T)):
    # Scale vector length by sqrt(eigenvalue) for visualization
    vec_scaled = vec * np.sqrt(val) * 2
    ax1.quiver(0, 0, vec_scaled[0], vec_scaled[1], angles='xy', scale_units='xy', scale=1,
               color='red' if i==0 else 'green', label=f'PC{i+1} Axis (Var={val:.2f})')

ax1.set_title("Centered Data with Principal Components")
ax1.set_xlim(-6, 6)
ax1.set_ylim(-6, 6)
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend()
ax1.set_aspect('equal')

# Reconstructed Data (1D Projection brought back to 2D space)
X_reconstructed = X_projected_1d @ W2.T
ax2.scatter(X_centered[:, 0], X_centered[:, 1], alpha=0.2, color='blue', label='Original Data')
ax2.scatter(X_reconstructed[:, 0], X_reconstructed[:, 1], alpha=0.6, color='red', label='1D Projected Data')
ax2.set_title("Dimensionality Reduction: 2D Projected onto 1D PC1")
ax2.set_xlim(-6, 6)
ax2.set_ylim(-6, 6)
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend()
ax2.set_aspect('equal')

plt.tight_layout()
plt.show()
```
